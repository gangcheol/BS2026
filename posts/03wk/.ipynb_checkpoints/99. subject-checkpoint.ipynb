## 1. 연구배경 (문제 상황 및 분석 목적)

`1` 본 연구는 흑색종(Melanoma) 환자의 조직생검에서 얻은 두 세포 집단인 **림프구 세포(Lymphocyte, $n_1=40$)** 와 **종양 세포(Tumor, $n_2=40$)** 의 직경(단위: $\mu\text{m}$) 데이터를 비교·분석한다.

`2` 주요 목적은 림프구 세포의 크기가 종양 세포의 크기보다 통계적으로 유의하게 작다고 볼 수 있는지 $5\%$ 유의수준에서 가설검정을 수행하는 것이다.

### 연구질문

**‘림프구세포의 크기가 종양세포의 크기보다 작다고 볼 수 있는가?**

## 2. 가설설정 (귀무가설/대립가설)

`1` 통계적 모형

 $$y_{ij} = \mu + \tau_i + \varepsilon_{ij},\quad (i = 1: \text{Lymphocyte},\,\,2: \text{Tumor}; \,\,j = 1, \dots, 40)$$

<details>
    <summary>기호 설명</summary>

- $y_{ij}$ (반응변수 / 관측값)
    - $i$번째 세포 집단(1: 림프구, 2: 종양)에 속한 $j$번째 개별 세포의 크기(직경, $\mu\text{m}$)
 
- $\mu$ (전체평균, Overall Mean)
    - 두 세포 집단 전체를 아우르는 기준 모평균값

- $\tau_i$ (집단효과, Group Effect): $i$번째 세포 종류가 세포 크기에 미치는 고유한 효과(처리 효과)
    - $\tau_1$: 림프구 세포(Lymphocyte) 집단의 효과 ($\mu_1 = \mu + \tau_1$)
    - $\tau_2$: 종양 세포(Tumor) 집단의 효과 ($\mu_2 = \mu + \tau_2$)

- $\varepsilon_{ij}$ (오차항, Random Error): 모형으로 설명되지 않는 $i$번째 집단 내 $j$번째 관측치의 확률적 오차
    - *(가정: 오차항은 서로 독립이고 평균이 0인 정규분포 $\varepsilon_{ij} \sim N(0, \sigma_i^2)$를 따름)*

- $i$ (집단 인덱스): 세포의 종류를 나타내는 첨자 ($i = 1$: 림프구 세포, $i = 2$: 종양 세포)

- $j$ (관측치 인덱스): 각 세포 집단 내 개별 세포 관측치의 번호를 나타내는 첨자 ($j = 1, 2, \dots, 40$)

</details>

`2` 모평균 정의

- $\mu_1$: 림프구 세포 크기(직경)의 모평균
- $\mu_2$: 종양 세포 크기(직경)의 모평균

`3` 검정가설 (단측검정, One-sided Test)

- 귀무가설 ($H_0$): $\mu_1 \ge \mu_2$ (림프구 세포의 평균 크기가 종양 세포의 평균 크기보다 크거나 같다)
- 대립가설 ($H_1$): $\mu_1 < \mu_2$ (림프구 세포의 평균 크기가 종양 세포의 평균 크기보다 작다)

### 전제조건 점검

SAS `PROC TTEST` 산출 결과는 다음과 같다

| 전제조건 | 점검방법 | 통계량 | p 값 | 판정 |
|:---:|:---:|:---:|:---:|:---:|
| 정규성(림프구세포) | Shapiro-Wilk | W = 0.98 | 0.6046 | 정규성 기각 못 함 |
| 정규성(종양세포) | Shapiro-Wilk | W = 0.98 | 0.7092 | 정규성 기각 못 함 |
| 등분산성 |  F-test | F = 3.2405 | 0.0004 | 등분산성 기각 |
| 등분산성 |  Levene's test | F = 444.73 | <.0001 | 등분산성 기각 |

###  SAS 코드

```python

DATA cell;
 INPUT group $ diameter @@;
 DATALINES;
L 9.0 L 9.4 L 4.7 L 4.8 L 8.9 L 4.9 L 8.4 L 5.9 L 6.3 L 5.7
L 5.0 L 3.5 L 7.8 L 10.4 L 8.0 L 8.0 L 8.6 L 7.0 L 6.8 L 7.1
T 12.6 T 14.6 T 16.2 T 23.9 T 23.3 T 17.1 T 20.0 T 21.0 T 19.1 T 19.4
T 16.7 T 15.9 T 15.8 T 16.0 T 17.9 T 13.4 T 19.1 T 16.6 T 18.9 T 18.7
;
RUN;
/* (1) 등분산 검정과 독립 2-표본 t-검정 */
PROC TTEST DATA=cell;
 CLASS group; /* 집단 구분 변수 */
 VAR diameter; /* 반응변수 */
RUN;

/* (2) 정규성 점검 (Shapiro-Wilk) */
PROC UNIVARIATE DATA=cell NORMAL;
 CLASS group;
 VAR diameter;
RUN;

PROC GLM DATA=ex3_2;
    CLASS group;
    MODEL x = group;
    MEANS group / HOVTEST=LEVENE;
RUN;


```

:::

## 4. 분석/모형적합

`1` 두 독립집단 데이터가 정규성 가정은 만족하지만 등분산 가정이 기각되었으므로,

`2` **Satterthwaite 근사법을 적용한 이분산 독립 2표본 t-검정(Satterthwaite/Welch Two-sample t-test)을 선택** 

`3` 보완적으로 비모수 검정인 Wilcoxon 순위합 검정도 병행 확인 $\to$ (소표본이므로 `Robust함을 확보하기 위함.`)

## 5. 결과해석 - 적합결과 및 검정통계량 해석

SAS 프로그램에서 `PROC TTEST` 산출 결과는 다음과 같다.

`1` **등분산 F-검정 (Equality of Variances)**
- $F = 3.24, \; df=(39, 39), \; p = 0.0004$ (이분산성)

`2` Satterthwaite t-검정 (SIDES=L, 이분산 적용)

$$t = \frac{(\bar{X} - \bar{Y}) - \delta}{\sqrt{s_1^2/n_1 + s_2^2/n_2}}, \qquad \nu = \frac{(w_1 + w_2)^2}{w_1^2/(n_1 - 1) + w_2^2/(n_2 - 1)}, \qquad w_i = \frac{s_i^2}{n_i}$$

<details>
    <summary>수식 대입 결과</summary>

$$w_1 = \frac{1.596^2}{40} = 0.0637, \quad w_2 = \frac{2.873^2}{40} = 0.2063 \quad \Rightarrow \quad \nu = \frac{(0.0637 + 0.2063)^2}{0.0637^2/39 + 0.2063^2/39} \approx 60.98$$ $$t = \frac{6.950 - 17.908}{\sqrt{0.0637 + 0.2063}} \approx -21.09$$ **결론**: 귀무가설 기각 ($p < 0.0001$) — 림프구 세포의 크기가 종양 세포보다 통계적으로 유의하게 작음. *(참고: 자유도가 60.98로 정수가 아닌 이유는 두 표본의 분산 차이와 정보량을 비대칭적으로 반영하기 때문)*

</details>

`3` 비모수 Wilcoxon 순위합 검정 결과

<details>
    <summary>Wilcoxon 순위합 검정 코드</summary>

```python
/* 3. 비모수 검정 (Wilcoxon 순위합 검정 / Mann-Whitney U test) */
PROC NPAR1WAY DATA=ex3_2 WILCOXON;
    CLASS group;
    VAR x;
RUN;
```

</details>

`1` 본 자료는 정규성 가정을 만족하나 표본의 수가 적어  `Robust`함을 검증하기 위해 정규성 가정이 깨지고, 이분산일 경우 적용하는 `Wilcoxon 순위합 검정`을 함께 살펴봄.

`2` 검정결과 정규근사 $Z = -7.66, \; p < 0.0001$으로 단측 $p$-값이 유의수준 $\alpha = 0.05$보다 현저히 작으므로($p < 0.0001$), 귀무가설 $H_0$를 강력히 기각.

`3` **즉, 림프구 세포의 평균 크기가 종양 세포의 평균 크기보다 전반적으로 더 낮은 경향이 있었다.**

## 6. 결론

::: {.panel-tabset}

##### 2-(1)

`1` 독립 2-표본 t-검정을 이용하여 림프구 세포와 종양 세포의 크기(직경) 평균을 비교한 결과, 

`2` 림프구 세포($n_1=40$)의 평균은 $6.95\,\mu\text{m}\,(SD=1.60)$, 종양 세포($n_2=40$)의 평균은 $17.91\,\mu\text{m}\,(SD=2.87)$이었다.

`3` 등분산성 검정 결과 $F = 3.24, p = 0.0004$로 등분산 가정이 기각되어 Satterthwaite 근사법을 적용.

`4`  검정 결과 $t(60.98) = -21.09$, 단측 $p < 0.0001$로 통계적으로 매우 유의한 차이가 있었음.

`5` 따라서 유의수준 5%에서 림프구 세포의 **평균** 크기가 종양 세포의 **평균** 크기보다 **통계적으로 유의하게 작다고 결론지을 수 있다.**

##### 2-(2)

독립 2표본 t-검정(Independent two-sample t-test)을 수행하기 위해 필요한 기본 가정은 다음과 같다

`1` **독립성 가정 (Independence)**: 두 집단(림프구 세포, 종양 세포) 내의 관측치들이 서로 독립적이어야 하며, 두 집단 간에도 서로 독립적으로 추출되어야 한다.

`2` **정규성 가정 (Normality)**: 각 집단에서 측정된 반응변수(세포 크기)의 모집단 분포가 정규분포를 따라야 한다
- (본 데이터는 $n_1=40, n_2=40$으로 충분히 크고 Shapiro-Wilk 검정에서 $p > 0.05$로 정규성을 만족함)

`3` **등분산성 가정 (Homogeneity of Variance)**: 두 모집단의 분산이 동일해야 한다는 가정이다[4].
* *참고*: 등분산성 가정이 성립하면 **합동분산(Pooled) t-검정**을 사용하지만,
* 본 데이터처럼 등분산 가정이 기각되는 경우에는 자유도를 보정한 **Satterthwaite(Welch) t-검정**을 적용하여 분석해야 한다

다음은 정신지체아를 둔 11명의 부모를 대상으로 아이들의 자립심을 길러주기 위한 교육을 전문가들로부터 받기 전과 받은 후의 지식 습득 점수를 측정한 자료이다.

<details>
    <summary>자료 생성 코드</summary>

```python
DATA ex3_4;
	INPUT id before after @@; /* id: 부모 번호, before: 사전 점수, after: 사후 점수 */
	diff = after - before; /* 점수 차이 변수 생성 (사후 - 사전) */
	DATALINES;
1 104 108 2 116 118 3 84 89 4 77 71 5 61 66
6 84 83 7 81 88 8 72 76 9 61 68 10 97 96 11 84 81
;
RUN;
```

</details>

| 부모 ID (\\(i\\)) | 사전 점수 ($y_{\text{before},i}$) | 사후 점수 ($y_{\text{after},i}$) | 점수 차이 ($d_i$) |
| :---: | :---: | :---: | :---: |
| **1** | 104 | 108 | +4 |
| **2** | 116 | 118 | +2 |
| **3** | 84 | 89 | +5 |
| **4** | 77 | 71 | -6 |
| **5** | 61 | 66 | +5 |
| **6** | 84 | 83 | -1 |
| **7** | 81 | 88 | +7 |
| **8** | 72 | 76 | +4 |
| **9** | 61 | 68 | +7 |
| **10** | 97 | 96 | -1 |
| **11** | 84 | 81 | -3 |
| 평균  | **81.909** | **84.000** | **+2.091** |

## 1. 연구배경

`1` 본 연구는 정신지체아를 둔 11명의 부모($n=11$)를 대상으로 자립심 향상 교육 실시 전(사전 점수, Before)과 실시 후(사후 점수, After)의 지식 습득 점수를 비교·분석한다.

`2` 동일한 연구 대상으로부터 처방(교육) 전후로 연속형 반응변수를 반복 측정한 짝지어진 자료(Paired data)이므로,

`3` 교육을 받기 전과 받은 후 지식 습득 점수에 통계적으로 유의한 차이가 있는지를 $5\%$ 유의수준에서 검정하고,

`4` 점수 차이 모평균의 $95\%$ 신뢰구간을 산출하는 것을 목적으로 한다.

### 연구질문

**교육 전후에 부모의 지식습득점수에 차이가 있는가?**

## 2. 가설설정

`1` 통계적 모형

$$d_i = y_{\text{after}, i} - y_{\text{before}, i} \quad (i = 1, 2, \dots, 11), \quad d_1, d_2, \dots, d_{11} \overset{iid}{\sim} N(\mu_d, \sigma_d^2)$$

<details>
    <summary>기호 설명</summary>

- $d_i$: $i$번째 부모의 교육 전후 지식 습득 점수 차이
- $y_{\text{after}, i}$: 교육 후 지식 습득 점수
- $y_{\text{before}, i}$: 교육 전 지식 습득 점수
- $\mu_d$: 교육 전후 점수 차이($\text{사후 점수} - \text{사전 점수}$)의 모평균
- $\sigma_d^{2}$: 교육 전후 점수 차이의 모분산

</details>

`2` 검정가설(양측검정, Two-sided Test)

- 귀무가설 ($H_0$): $\mu_d = 0$
    - 교육 전과 후의 지식 습득 점수 `평균에 차이가 없다`
- 대립가설 ($H_A$): $\mu_d \neq 0$
    - 교육 전과 후의 지식 습득 점수 `평균에 차이가 있다`

## 3. EDA

## 4. 분석/모형적합

`1` 차이값 $d$에 대한 정규성이 충족하므로 **모수적** 방법인 대응표본 `t-검정(Paired t-test)`을 적용

- $W = 0.9170, \; p = 0.2941$
- $p > 0.05$이므로 차이값 분포가 정규성을 만족한다는 가정을 충족

<details>
    <summary>SAS 실행 코드</summary>

```SAS
/* PROC TTEST의 PAIRED 문 사용*/
PROC TTEST DATA=ex3_4 ALPHA=0.05;
	PAIRED after*before; /* after - before 차이에 대한 t-검정 및 95% 신뢰구간 산출 */
RUN;
```

</details>

## 5. 결과해석

| t 통계량 (df = 10) | 1.62 | 양측 $p = 0.1360$ |

## 6. 결론

::: {.panel-tabset}

##### 4-(1)

`1` 유의수준 $\alpha = 0.05$하에서 대응표본 t-검정을 실시한 결과, 검정통계량 $t(10) = 1.62$, 양측 $p\text{-value} = 0.1360$으로 나타남.

`2` 따라서 $p$-값이 유의수준 $0.05$보다 크므로($p > 0.05$) 귀무가설 $H_0$를 기각할 수 없다.

`3` **즉, 교육 전과 후 부모들의 지식 습득 점수에는 통계적으로 유의한 차이가 있다고 볼 수 없다.**

##### 4-(3)

`2` **(2)의 신뢰구간**: 산출된 95% 신뢰구간 $[-0.7823, 4.9641]$ 내에 귀무가설의 기준값인 '0'이 포함되어 있다.

`3` **결론**:
    
-  귀무가설 $\mu_d = 0$을 유의수준 5%에서 기각할 수 없음을 의미하므로,
-  **(1)의 가설검정 결과와 (2)의 신뢰구간 결과는 완전하게 일치하고 연관**되어 있음을 의미한다!

## 1. 연구배경

`1` 생활사건이 유발하는 심리적 스트레스가 면역기능에 영향을 미치는지 밝히는 것은 심신의학(Psychosomatic Medicine)의 핵심 연구 주제이다

`2` 본 연구는 전이성 폐암(Metastatic lung cancer) 환자 36명($N=36$)을 대상으로 최근 생활 스트레스 총량을 반영하는 SRRS(Social Readjustment Rating Scale) 점수를 측정하여

`3` 저($\le 54$), 중($55 \sim 99$), 고($\ge 100$) 세 집단(각 $n_i = 12$)으로 분류하고, 면역기능 지표인 NK 세포활동(Natural Killer cell activity) 수치를 측정하였다.

**연구질문**: **SRRS 스트레스 수준(저, 중, 고)에 따라 환자들의 NK 세포활동 평균에 차이가 있는가? 차이가 있다면 어느 집단 사이에 존재하는가?**

## 2. 가설설정

`1` 통계적 모형

$$y_{ij} = \mu + \tau_i + \varepsilon_{ij} \quad (i = 1: \text{저}, \; 2: \text{중}, \; 3: \text{고}; \quad j = 1, \dots, 12), \quad \varepsilon_{ij} \overset{iid}{\sim} N(0, \sigma^2)$$

<details>
    <summary>기호 설명</summary>

* **$\mu$**: 전체 모평균 (Grand population mean)

* **$\tau_i$**: i번째 SRRS 집단의 효과 (Group effect, $i=1$: 저, $2$: 중, $3$: 고)
 
* **$\varepsilon_{ij}$**: i번째 집단의 j번째 관측치에 대한 오차항 (Random error)

</details>

`2`  가설설정

$$H_0: \mu_1 = \mu_2 = \mu_3 \quad \text{vs} \quad H_1: \text{not }H_0$$

<details>
    <summary>기호 설명</summary>

* **$\mu_1, \mu_2, \mu_3$**: 각각 저, 중, 고 SRRS 집단의 NK 세포활동 모평균

- $H_0$: 세 스트레스 집단의 `NK 세포활동 모평균이 모두 같다`.

- $H_1$: 적어도 한 집단의 `NK 세포활동 모평균은 다른 집단과 다르다`

</details>

### 전제조건 점검(정규성 & 등분산)

| 전제조건 | 점검방법 | 통계량 | p 값 | 판정 |
|---|---|---|---|---|
| 정규성(저) | Shapiro-Wilk | W = 0.90 | 0.1555 | 기각 못 함 |
| 정규성(중) | Shapiro-Wilk | W = 0.89 | 0.1019 | 기각 못 함 |
| 정규성(고) | Shapiro-Wilk | W = 0.89 | 0.1243 | 기각 못 함 |
| 등분산성 | Levene(평균 중심) | F = 6.39 | 0.0045 | 등분산 기각 |
| 등분산성 | Bartlett | $\chi^2 = 22.01$ | <0.0001 | 등분산 기각 |

`1` **Shapiro-Wilk 정규성 검정**: 세 집단 모두 유의수준 5%에서 귀무가설(정규분포를 따른다)을 기각하지 못하여 **정규성 가정을 충족**

`2` 등분산성 검정 (Levene / Bartlett Test)
- 저 집단의 큰 분산($s^2 = 720.43$)으로 인해 등분산 가정이 기각되었다.
- 따라서 `표준 일원분산분석표`와 함께 등분산 위반에 강건한 **Welch 분산분석** 및 비모수 **Kruskal-Wallis 검정**을 교차 검증으로 제시한다.

### EDA 분석 코드

```python
/******************************************************************
 * 1. 세 그룹 오버레이 히스토그램 (단일 그래프)
 ******************************************************************/
PROC SGPLOT DATA=ex3_5;
    HISTOGRAM activity / GROUP=group TRANSPARENCY=0.4; /* 그룹별 색상 구분 및 반투명 처리 */
    DENSITY activity / GROUP=group TYPE=NORMAL;        /* 그룹별 정규곡선 오버레이 */
    XAXIS LABEL="NK 세포활동 수치";
    YAXIS LABEL="비율 (%)";
    TITLE "SRRS 그룹별 NK 세포활동 오버레이 히스토그램";
RUN;

/******************************************************************
 * 2. 그룹별 색상 구분 상자그림 (단일 그래프)
 ******************************************************************/
/* CATEGORY=group GROUP=group 옵션을 함께 지정하면, x축에 세 그룹이 배치되면서 각각 고유한 색상(범례 포함)으로 채워집니다. */
PROC SGPLOT DATA=ex3_5;
    VBOX activity / CATEGORY=group GROUP=group; /* 그룹별 색상 구분 상자그림 */
    XAXIS LABEL="SRRS 스트레스 수준";
    YAXIS LABEL="NK 세포활동 수치" GRID;
    TITLE "SRRS 그룹별 NK 세포활동 상자그림";
RUN;

/******************************************************************
 * 3. 오버레이 Q-Q 플롯 (단일 그래프)
 ******************************************************************/
/* 1단계: PROC RANK를 통한 그룹별 이론적 정규 분위수(z_score) 산출 */
PROC SORT DATA=ex3_5;
    BY group;
RUN;

PROC RANK DATA=ex3_5 NORMAL=BLOM OUT=ex3_5_qq;
    BY group;
    VAR activity;
    RANKS z_score;
RUN;

/* 2단계: 단일 그래프에 세 그룹의 Q-Q 관측점 및 기준선 오버레이 */
PROC SGPLOT DATA=ex3_5_qq;
    SCATTER X=z_score Y=activity / GROUP=group;               /* 그룹별 색상 산점도 */
    REG X=z_score Y=activity / GROUP=group NOMARKERS;         /* 그룹별 기준 직선 */
    XAXIS LABEL="이론적 정규 분위수 (Theoretical Quantiles)";
    YAXIS LABEL="관측된 NK 세포활동 수치 (Sample Quantiles)" GRID;
    TITLE "SRRS 그룹별 NK 세포활동 오버레이 Q-Q 플롯";
RUN;

/******************************************************************
 * 참고: 1행 3열 가로 배치 패널 Q-Q 플롯
 ******************************************************************/
/* Q-Q 플롯을 한 그래프에 겹치는 것보다 가로 1줄로 나누어 보고 싶을 때 사용 */
PROC SGPANEL DATA=ex3_5_qq;
    PANELBY group / ROWS=1 COLUMNS=3;
    SCATTER X=z_score Y=activity / GROUP=group;
    REG X=z_score Y=activity / GROUP=group NOMARKERS;
    COLAXIS LABEL="이론적 분위수";
    ROWAXIS LABEL="관측 수치";
RUN;
```

## 4. 분석/모형적합 & 결과해석 (소문제(2), (3)번 답변)

::: {.panel-tabset}

### **ANOVA Table**

| 변동요인 (Source) | 제곱합 (SS) | 자유도 (df) | 평균제곱 (MS) | F 통계량 | p-value |
|---|---|---|---|---|---|
| 집단 간 | 4,506.11 | 2 | 2,253.06 | 7.85 | 0.0016 |
| 집단 내 | 9,470.72 | 33 | 286.99 | - | - |
| 합계 | 13,976.83 | 35 | - | - | - |

$$F = \frac{MSTr}{MSE} = \frac{2,253.06}{286.99} = 7.85 \quad (df_1 = 2, \; df_2 = 33)$$

`1` F 통계량 및 p-value: $F(2, 33) = 7.85, \; p = 0.0016 < 0.05$

`2` $p$-값이 유의수준 $\alpha = 0.05$보다 현저히 작으므로 귀무가설 $H_0$를 기각한다.

- 즉, SRRS 스트레스 수준에 따라 환자의 NK 세포활동 평균에는 통계적으로 유의한 차이가 존재한다.

### Tukey HSD 다중비교

| 비교 집단 | 평균차 | $p\text{-adj}$ | 판정 |
|---|---|---|---|
| 저 집단 vs 중 집단 | 24.87 | **0.0029** | 통계적으로 유의함 |
| 저 집단 vs 고 집단 | 22.41 | **0.0075** | 통계적으로 유의함 |
| 중 집단 vs 고 집단 | -2.46 | **0.9329** | 유의한 차이 없음 |

### 보완 검정

* **Welch 분산분석**: $F(2, 20.3) = 6.22, \; p = 0.0078 < 0.05$
* **Kruskal-Wallis 비모수 검정**: $\chi^2(2) = 9.77, \; p = 0.0076 < 0.05$

> **해석**: 등분산 가정에 대한 대안 방법으로 동일하게 유의수준 5%에서 집단 간 유의한 차이가 입증되어 분석 결론이 매우 안정적이다.

### SAS 코드

```python
/* (1) 일원분산분석 + 등분산 검정 + Tukey 다중비교 */
PROC GLM DATA=nk;
 CLASS group;
 MODEL activity = group;
 MEANS group / HOVTEST=LEVENE TUKEY;
RUN;
QUIT;
/* (2) 정규성 점검: 잔차 저장 후 Shapiro-Wilk */
PROC GLM DATA=nk;
 CLASS group; MODEL activity = group;
 OUTPUT OUT=out R=resid;
RUN;
QUIT;
PROC UNIVARIATE DATA=out NORMAL;
 VAR resid;
RUN;
/* (3) 비모수 대안 */
PROC NPAR1WAY DATA=nk WILCOXON KRUSKALWALLIS;
 CLASS group; VAR activity;
RUN;
```

:::

## 5. 결론

`1` 일원분산분석(모형 $y_{ij} = \mu + \tau_i + \varepsilon_{ij}$)을 이용하여 SRRS 저·중·고 세 집단 환자의 NK 세포활동을 비교하였다.

`2` 세 집단의 평균은 각각 저 집단 $40.47 \pm 26.84$, 중 집단 $15.60 \pm 6.42$, **고 집단 $18.06 \pm 9.97$**이었다.

`3` 분산분석 결과 $F(2, 33) = 7.85, \; p = 0.0016$으로 유의수준 5%하에서 통계적으로 유의한 차이가 있었고,

`4` Tukey 다중비교 결과, 스트레스가 가장 낮은 '저' 집단의 NK 세포활동 수치가 '중' 집단($p = 0.0029$) 및 '고' 집단($p = 0.0075$)보다 통계적으로 유의하게 높았으며,

`5` '중' 집단과 '고' 집단 사이에는 유의한 차이가 없었다($p = 0.9329$).

`6` 따라서 생활 스트레스 수치가 낮은 환자일수록 면역기능 지표인 NK 세포활동 능력이 평균적으로 높게 유지됨이 통계적으로 유의미하다는 것을 입증했다.

# 7번

식이요법과 운동방법이 체중 감량에 미치는 영향분석

| 운동 / 식이 | I | II | III | IV |
|:---:|:---:|:---:|:---:|:---:|
| A | 7.0 | 5.3 | 4.9 | 8.8 |
| B | 9.9 | 5.7 | 7.6 | 8.9 |
| C | 8.5 | 4.7 | 5.5 | 8.1 |
| D | 5.1 | 3.5 | 2.8 | 3.3 |
| E | 10.3 | 7.7 | 8.4 | 9.1 |

## 1. 연구배경

다이어트 연구자가 20명의 참가지(`N=20`)를 대상으로 **'운동방법 5가지(A, B, C, D, E)'** 와 **'식이요법 4가지(Ⅰ, Ⅱ, Ⅲ, Ⅳ)'** 의 조합(총 20개 셀)에 각 1명씩 임의 배정하고, 3주간 처치 후 체중 감량(kg)을 측정하였다.

### 연구질문

`1` 운동방법에 따라 체중 감량 효과에 통계적으로 유의한 차이가 있는가?

`2` 식이요법에 따라 체중 감량 효과에 통계적으로 유의한 차이가 있는가?

`3` 차이가 존재한다면, 어느 운동방법 및 식이요법 수준 사이에서 유의한 차이가 나타나는가?

::: {.callout-important}

1. 셀(각 조합별)당 관측치가 1개(`n=1`)이므로 오차항과 교호작용항을 분리할 수 없다. 따라서 두 요인이 서로 독립적으로 작용한다는 **가법모형(Additive Model)** 을 적용하며,

2. 분석 전 **Tukey 가법성 검정(Tukey's Test of Additivity)** 을 통해 교호작용 부재 가정을 검증한다.

:::

## 2. 가설설정

운동방법 $i$의 효과를 $\alpha_i \; (i=1,\dots,5)$, 식이요법 $j$의 효과를 $\beta_j \; (j=1,\dots,4)$라 할 때, 검정가설은 다음과 같다.

`1` 통계적 모형

$$y_{ij} = \mu + \alpha_i + \beta_j + \varepsilon_{ij} \quad (i = 1,\dots,5; \quad j = 1,\dots,4), \quad \varepsilon_{ij} \overset{iid}{\sim} N(0, \sigma^2)$$

<details>
    <summary>기호</summary>

* **$y_{ij}$**: $i$번째 운동방법과 $j$번째 식이요법을 적용한 체중 감량 관측치

* **$\mu$**: 전체 평균 (Grand mean)

* **$\alpha_i$**: $i$번째 요인 A(운동방법)의 주효과

* **$\beta_j$**: $j$번째 요인 B(식이요법)의 주효과

* **$\varepsilon_{ij}$**: 확률오차 (서로 독립이며 평균이 0, 분산이 $\sigma^2$인 정규분포를 따름)

</details>

`2` 가설 1 (운동방법의 주 효과)

- 귀무가설 ($H_0$): $\alpha_1 = \alpha_2 = \alpha_3 = \alpha_4 = \alpha_5 = 0$ (운동방법에 따른 체중 감량 차이가 없다)

- 대립가설 ($H_1$): 적어도 하나의 $\alpha_i \neq 0$ (적어도 한 운동방법의 효과가 다르다)

`3` 가설 2 (식이요법의 주효과)

- 귀무가설 ($H_0$): $\beta_1 = \beta_2 = \beta_3 = \beta_4 = 0$ (식이요법에 따른 체중 감량 차이가 없다)

- 대립가설 ($H_1$): 적어도 하나의 $\beta_j \neq 0$ (적어도 한 식이요법의 효과가 다르다)

### 가정 검정 결과

| 전제조건 | 점검 방법 | 통계량 | p-value | 최종 판정 |
|---|---|---|---|---|
| 잔차 정규성 | Shapiro-Wilk Test | $W = 0.96$ | 0.61 | 정규성 충족 ($p > 0.05$) |
| 가법성 (교호작용 부재) | Tukey's Additivity Test | $F = 0.87$ | 0.37 | 가법성 충족 ($p > 0.05$) |
| 등분산 | Levene | $F = 1.33$ | 0.30 |  등분산 충족  ($p > 0.05$) |

`1` **Shapiro-Wilk 검정을 통해 잔차의 정규성 가정을 확인**하였으며, 각 처리 조합당 관측치가 1개인 반복 없는 데이터이므로 **Tukey 가법성 검정을 통해 두 요인의 상호작용(교호작용) 여부를 검정**

`2` 잔차의 정규성 검정 결과 정규성 가정을 충족하였다. ($W=0.963, p=0.612$)

`3` 또한 운동방법×식이요법 교호작용(가법성)은 유의하지 않았다. ($F=0.87, p=0.372$).

`4` 따라서 두 요인이 서로 독립적으로 작동함을 확인하였으며, **교호작용항을 제외한 가법 이원분산분석(Additive Two-way ANOVA)을 적용**하여 각 요인의 주효과를 분석하였다.

### SAS 코드

```python
/******************************************************************
 * 문제 7번: 이원배치 정석 EDA (교호작용 플롯 + 잔차 정규성 진단)
 ******************************************************************/
 
/* 1. 데이터 입력 */
DATA diet;
    INPUT exercise $ diet $ loss @@;
    DATALINES;
A I 7.0 A II 5.3 A III 4.9 A IV 8.8
B I 9.9 B II 5.7 B III 7.6 B IV 8.9
C I 8.5 C II 4.7 C III 5.5 C IV 8.1
D I 5.1 D II 3.5 D III 2.8 D IV 3.3
E I 10.3 E II 7.7 E III 8.4 E IV 9.1
;
RUN;

/* 2. [핵심 EDA 1] 교호작용 프로필 플롯 */
PROC SGPLOT DATA=diet;
    SERIES X=diet Y=loss / GROUP=exercise MARKERS LINEATTRS=(THICKNESS=2);
    XAXIS LABEL="식이요법 (Diet)";
    YAXIS LABEL="평균 체중 감량 (kg)" GRID;
    TITLE "운동방법 x 식이요법 교호작용 프로필 플롯";
RUN;

/* 3. 모형 적합 후 잔차(resid) 추출 */
PROC GLM DATA=diet NOPRINT;
    CLASS exercise diet;
    MODEL loss = exercise diet;
    OUTPUT OUT=diag_out R=resid; /* 잔차 저장 */
RUN;
QUIT;

/* 4. [핵심 EDA 2 & 3] 잔차 히스토그램 및 잔차 Q-Q 플롯 */
PROC UNIVARIATE DATA=diag_out NORMAL PLOT;
    VAR resid;
    HISTOGRAM resid / NORMAL;
    QQPLOT resid / NORMAL(MU=EST SIGMA=EST);
RUN;

/* 5.  가법성 검정 */
/* 1단계: 기본 가법모형 적합 및 적합치(pred) 저장 */
PROC GLM DATA=diet NOPRINT;
    CLASS exercise diet;
    MODEL loss = exercise diet;
    OUTPUT OUT=diet_pred P=pred; /* 적합치(y_hat)를 pred라는 변수로 출력 */
RUN;
QUIT;

/* 2단계: DATA 스텝에서 적합치(pred)의 제곱항(pred2) 생성 */
DATA diet_pred;
    SET diet_pred;
    pred2 = pred * pred; /* pred^2 산출 */
RUN;

/* 3단계: 제곱항(pred2)을 추가하여 Tukey 가법성 검정 F통계량 산출 */
PROC GLM DATA=diet_pred;
    CLASS exercise diet;
    MODEL loss = exercise diet pred2 / SS3; /* Type III SS 확인 */
RUN;
QUIT;
```

`1` 각 조합의 관측치가 1개이고, 실제 교호작용 검정결과가 독립이었으므로 모형에 교호작용항을 넣지 않는다. 

### 분산분석표

`2` 식이요법 주효과: $F(3, 12) = 11.83, \; p = 0.0007 < 0.05$로 귀무가설 기각.
- 식이요법 종류에 따라 체중 감량 평균에 통계적으로 유의한 차이가 존재함.


`3` 또한 두 요인이 함께 설명하는 전체 체중 감량 변동은 `(62.65+27.43)/99.35`인 `90.7%`에 이른다.

$$R^{2} = \frac{SS_{\text{운동방법}}+SS_{\text{식이요법}}}{\text{Total}} = \frac{62.65+27.43}{99.35} = 90.67\%$$

### SCHEFFE

<details>
    <summary>사전 지식</summary>

앞선 분산분석표를 살펴본 결과, 각각의 방법에 따라 체중 감량 차이가 통계적으로 유의미했다. 따라서 구체적으로 어느 방법에서 체중감량 차이가 유의미한지 확인하기 위해, `SCHEFFE` 다중비교를 수행하였다.

`1` ANOVA 분석의 한계인 **"정확히 몇 번 운동이 몇 번 운동보다 더 좋은지"**, **"몇 번 식단이 몇 번 식단보다 효과가 좋은지"** 1대 1로 콕 짚어주지는 못함.

`2` 그렇다고 A-B, A-C, A-D... 이렇게 짝을 지어 단순 t-검정을 여러 번 반복하면, 우연히 차이가 난 것을 진짜 차이라고 잘못 판단할 `확률(1종 오류)`이 엄청나게 커짐

`3` Scheffé(셰페) 사후검정은 여러 그룹을 일일이 짝지어 비교할 때 발생하는 **오류 확률을 엄격하게 통제해 주는 가장 안전하고 보수적인 통계 검정 방법**을 사용

</details>

**Scheffé 다중비교 유의한 결과 요약**

| 요인 | 비교 집단 | 평균차 (kg) | F 통계량 | Scheffé 임계값 | p-value | 최종 판정 |
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| 운동 | B vs D | 4.35 | 48.95 | 13.04 | 0.0003 | 유의함 (B > D) |
| 운동 | D vs E | -5.20 | 69.95 | 13.04 | 0.0001 | 유의함 (E > D) |
| 운동 | A vs D | 2.83 | 20.65 | 13.04 | 0.0118 | 유의함 (A > D) |
| 운동 | C vs D | 3.03 | 23.67 | 13.04 | 0.0072 | 유의함 (C > D) |
| 운동 | A vs E | -2.38 | 14.59 | 13.04 | 0.0362 | 유의함 (E > A) |
| 식이 | Ⅰ vs Ⅱ | 2.78 | 24.99 | 10.47 | 0.0029 | 유의함 (Ⅰ > Ⅱ) |
| 식이 | Ⅰ vs Ⅲ | 2.32 | 17.41 | 10.47 | 0.0109 | 유의함 (Ⅰ > Ⅲ) |
| 식이 | Ⅱ vs Ⅳ | -2.26 | 16.52 | 10.47 | 0.0130 | 유의함 (Ⅳ > Ⅱ) |
| 식이 | Ⅲ vs Ⅳ | -1.80 | 10.48 | 10.47 | 0.0499 | 유의함|

`1` 운동방법 비교 (어떤 운동이 가장 효과적인가?)

- D 운동은 최악의 운동방법임
- D 운동과 다른 운동들을 비교했을 때(B vs D, D vs E, A vs D, C vs D), 전부 $p < 0.05$로 유의한 차이가 났다.
- 체중 감량 효과는 **E군(8.88kg)과 B군(8.03kg)이 최상위**이며, **D군(3.68kg)이 최하위**이다. 

`2` 식이요법 비교 (어떤 식단이 가장 효과적인가?)

- **식단 Ⅰ과 Ⅳ가 우수한 식단임**:
    - Ⅰ 요법(8.16kg)과 Ⅳ 요법(7.64kg)은 Ⅱ 요법(5.38kg) 및 Ⅲ 요법(5.84kg)에 비해 체중이 유의하게 더 많이 빠졌다.
 
- 체중 감량 효과는 **Ⅰ 요법(8.16kg)과 Ⅳ 요법(7.64kg)이 우수**하고, **Ⅱ 요법과 Ⅲ 요법은 효과가 떨어짐.

`3` 살빼려면?

- 체중 감량 효과를 보려면 운동방법은 `E나 B`로 식단은 `Ⅰ과 Ⅳ`를 하는 것이 좋음 

### SAS 코드

```python
PROC GLM DATA=diet;
    CLASS exercise diet;
    MODEL loss = exercise diet / SS3;
    MEANS exercise diet / SCHEFFE TUKEY LINES; /* Scheffé 및 Tukey 다중비교 */
    OUTPUT OUT=diag_out R=resid;               /* 잔차(resid) 저장 */
    TITLE "가법 이원분산분석 및 Scheffé 사후검정";
RUN;
QUIT;
```

## 5. 결론

`1` 이원배치 가법모형($y_{ij} = \mu + \alpha_i + \beta_j + \varepsilon_{ij}$)을 통해 식이요법 4가지와 운동방법 5가지의 체중 감량 효과를 검정함.

`2` EDA 분석에서는 교호작용 프로필 플롯의 평행성과 잔차의 정규성 및 Tukey 가법성($p=0.3716$)을 보였고,

`3` 운동방법($F=20.26, p<0.0001$)과 식이요법($F=11.83, p=0.0007$) 모두 체중 감량에 유의한 주효과가 있음을 확인하였음.

`4` 체중 감량을 최대로 이끌어내기 위해서는 운동방법 `E(또는 B)`와 식이요법 `Ⅰ(또는 Ⅳ)`의 조합을 적용하는 것이 가장 이상적임.

## 2. 가설설정

`1` 통계적 모형

나이 $i$의 주효과를 $\alpha_i \; (i=1,2,3)$, 마리화나 사용여부 $j$의 주효과를 $\beta_j \; (j=1,2,3)$, 나이와 마리화나의 교호작용(상호작용) 효과를 $(\alpha\beta)_{ij}$라 할 때, 반복이 있는 이원분산분석의 검정가설은 다음과 같다.

$$y_{ijk} = \mu + \alpha_i + \beta_j + (\alpha\beta)_{ij} + \varepsilon_{ijk}$$

 (단, $i=1,2,3; \; j=1,2,3; \; k=1,2,3$ 이며, 오차항 $\varepsilon_{ijk} \overset{iid}{\sim} N(0, \sigma^2)$)

<details>
    <summary>기호</summary>

* **$y_{ijk}$**: $i$번째 나이 집단과 $j$번째 마리화나 사용여부 조합의 $k$번째 감정성숙점수 관측치
 
* **$\mu$**: 전체 평균

* **$\alpha_i$**: $i$번째 나이 집단의 주효과


* **$\beta_j$**: $j$번째 마리화나 사용여부의 주효과


* **$(\alpha\beta)_{ij}$**: 두 요인 간의 교호작용 효과 (나이에 따라 마리화나 사용여부의 효과가 달라지는 정도)

* **$\varepsilon_{ijk}$**: 확률오차

</details>

::: {.panel-tabset}

### 가설1

나이 요인의 주효과 (Main Effect of Age)

`1` 귀무가설 ($H_0$): $\alpha_1 = \alpha_2 = \alpha_3 = 0$ (나이 집단 간 감정성숙점수 모평균에 차이가 없다)

`2` 대립가설 ($H_1$): 적어도 하나의 $\alpha_i \neq 0$ (적어도 한 나이 집단의 감정성숙점수 모평균은 다르다)

### 가설2

마리화나 사용여부 요인의 주효과 (Main Effect of Marijuana Use)

`1` 귀무가설 ($H_0$): $\beta_1 = \beta_2 = \beta_3 = 0$ (마리화나 사용여부에 따른 감정성숙점수 모평균에 차이가 없다)

`2` 대립가설 ($H_1$): 적어도 하나의 $\beta_j \neq 0$ (적어도 한 마리화나 사용 그룹의 감정성숙점수 모평균은 다르다)

### 가설3

 나이 $\times$ 마리화나 사용여부의 교호작용 (Interaction Effect)

`1` 귀무가설 ($H_0$): 모든 $(\alpha\beta)_{ij} = 0$ (나이와 마리화나 사용여부 간에 교호작용이 없다)

`2` 대립가설 ($H_1$): 적어도 하나의 $(\alpha\beta)_{ij} \neq 0$ (나이와 마리화나 사용여부 간에 교호작용이 존재한다)

:::

`2` 문제 9번 (반복 있음, $n=3$):

- 따라서 전제조건 단계에서는 **잔차 정규성**과 **등분산성** 2가지를 점검하고, 교호작용 존재 여부는 뒤이어 나오는 **이원분산분석표**에서 확인한다.

`1` 모형 설명력

**이원분산분석표**

`2` 분석 방향:

- 따라서 연령대(나이)별로 집단을 나누어 마리화나의 영향을 개별 검정하는 `단순주효과 분석(SLICE Analysis)`을 시행해야함.

### 단순주효과(SLICE) 분석

오차평균제곱($MSE = 9.22, \; df = 18$)을 기준으로 나이 집단별 마리화나 사용여부의 단순주효과를 F-검정(SLICE=age)한 결과

**나이 집단별 사용여부의 단순주효과**[cite: 4]

| 나이 집단 | 제곱합 (SS) | 자유도 (df) | 평균제곱 (MS) | F 통계량 | p-value | 판정 결과 | 그룹별 평균점수 (미사용 / 가끔 / 매일) |
|---|---|---|---|---|---|---|---|
| 15-19세 | $50.00$ | $(2, 18)$ | $25.00$ | $2.71$ | $0.0935$ | 유의하지 않음 ($p > 0.05$) | $25.00\text{점} \; / \; 20.00\text{점} \; / \; 20.00\text{점}$ |
| 20-24세 | $200.00$ | $(2, 18)$ | $100.00$ | $10.84$ | $0.0008$ | 유의함 ($p < 0.001$) | $30.00\text{점} \; / \; 20.00\text{점} \; / \; 20.00\text{점}$ |
| 25-29세 | $650.00$ | $(2, 18)$ | $325.00$ | $35.24$ | $< 0.0001$ | 매우 유의함 ($p < 0.001$) | $30.00\text{점} \; / \; 15.00\text{점} \; / \; 10.00\text{점}$ |

📝 연령대별 쉽게 푸는 결과 해설

`1` **15-19세 (청소년기 - 차이 미미함)**: $p = 0.0935 > 0.05$로 통계적으로 유의한 차이가 없었음.

- 10대 후반에는 마리화나 미사용자(25.0점)와 가끔/매일 사용자(각 20.0점) 간의 점수 차이가 유의수준 5% 내에서 확실한 차이라고 단정할 수 없었다.

- *(이 시기는 정서적 성숙이 한창 진행 중인 단계라 그룹 간 격차가 상대적으로 적게 나타남)*


`2` **20-24세 (초기 청년기 - 성숙도 격차 발생)**: $p = 0.0008 < 0.001$로 매우 유의한 차이를 보임.

- 20대 초반부터 마리화나 사용에 따른 감정성숙도 격차가 뚜렷해짐.

- 마리화나를 `사용하지 않는 그룹(30.0점)`에 비해, 가끔 또는 매일 사용하는 그룹(각 20.0점)은 감정성숙점수가 **10점 유의하게 낮았습니다.**

`3` **25-29세 (후기 청년기 - 극심한 발달 지체 및 폭락)**: $p < 0.0001$로 가장 강력한 통계적 유의성을 보임.

- 20대 중후반에서는 마리화나 사용 여부에 따른 감정성숙도 차이가 극심하게 벌어진다.

- `전혀 쓰지 않는 그룹(30.0점)`에 비해, **가끔 사용하는 그룹은 15점 낮은 15.0점**, **매일 사용하는 그룹은 무려 20점이나 폭락한 10.0점**을 기록했음.

### SAS 코드

```python
DATA maturity;
 INPUT age $ use $ score @@;
 DATALINES;
15-19 no 25 15-19 no 28 15-19 no 22
15-19 occ 18 15-19 occ 23 15-19 occ 19
15-19 day 17 15-19 day 24 15-19 day 19
20-24 no 28 20-24 no 32 20-24 no 30
20-24 occ 16 20-24 occ 24 20-24 occ 20
20-24 day 18 20-24 day 22 20-24 day 20
25-29 no 25 25-29 no 35 25-29 no 30
25-29 occ 14 25-29 occ 16 25-29 occ 15
25-29 day 10 25-29 day 8 25-29 day 12
;

RUN;
/* (1) 교호작용을 포함한 이원분산분석 */
PROC GLM DATA=maturity;
 CLASS age use;
 MODEL score = age use age*use / SS3;
 LSMEANS age*use / SLICE=age; /* 나이별 사용여부의 단순주효과 */
 MEANS age*use;
RUN;
QUIT;

/* (2) 전제조건 점검 */
PROC GLM DATA=maturity; CLASS age use;
 MODEL score = age use age*use; OUTPUT OUT=o R=r;
RUN;
QUIT;
PROC UNIVARIATE DATA=o NORMAL; VAR r; RUN;
```

:::

## 5. 결론

::: {.panel-tabset}

### 분석 결과 요약

`1` **전제조건 충족**:

- Shapiro-Wilk 잔차 정규성 검정($W=0.97, \; p=0.66$)과 Levene 등분산성 검정($F=0.57, \; p=0.79$)을 모두 충족하여 분산분석 모형의 통계적 타당성을 확보

`2` **교호작용의 유의성**:

- 나이와 마리화나 사용여부 간의 교호작용이 통계적으로 유의하였습니다 ($F(4, 18) = 4.97, \; p = 0.0071$).

- 전체 모형의 설명력($R^2$)은 `85.96%`로 매우 높았음.

::: {.callout-important}

💡 **연령대별 단순주효과(SLICE) 분석**:

`1` **15-19세 (청소년기)**:

- 마리화나 사용에 따른 감정성숙점수 차이가 `통계적으로 유의하지 않았다.` ($p = 0.0935$)

`2` **20-24세 (초기 청년기)**:

- 마리화나 사용에 따른 차이가 유의하게 나타났으며, 미사용자(30점)에 비해 가끔/매일 사용자(각 20점)의 성숙점수가 `10점 낮았다.` ($p = 0.0008$)

`3` **25-29세 (후기 청년기)**:

- 발달 격차가 가장 극심하게 벌어졌으며, 미사용자(30점) 대비 `가끔 사용자는 15점 낮은 15점`, `매일 사용자는 20점 폭락한 10점을 기록하였다.` ($p < 0.0001$)

:::

### 시사점

`1` **마리화나 영향의 연령 의존성**:

- 마리화나 사용이 감정성숙도 저하에 미치는 해로운 영향은 전 연령대에 동일하게 나타나는 것이 아니라,

- **나이가 들수록(20대 중후반으로 갈수록) 훨씬 치명적이고 심각하게 작용**한다.

`2` **정상 발달의 저해**:

- 마리화나를 쓰지 않는 청년층은 나이가 들면서 감정성숙도가 높게 유지되는 정상적 발달 경향을 보인 반면,

- 매일 사용하는 청년층은 20대 중후반에 감정성숙도가 급격히 퇴보하는 현상이 확인됨.

::: {.callout-important}

`3` **보건 정책적 제언**:

- 따라서 청년층(특히 20대) 대상의 마리화나 오남용 예방 프로그램 및 정서적 중재 정책이 시급히 마련될 필요가 있다!
