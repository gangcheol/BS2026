# 2번

다음은 흑색종(melanoma) 환자들의 조직생검으로부터 얻은 데이터로서 40개의 림프구와 40개의 종양세포의 직경에 대한 데이터이다.(단위: $\mu$m)

<center><img src = "연습문제 3-2번.jpg" width = 500></center>

## (1)

림프구세포의 크기가 조양세포의 크기보다 작다고 볼 수 있는가? 5% 유의수준으로 SAS를 이용하여 가설 검정하라. 

## (2)

위의 검정을 하기 위해 필요한 가정은 무엇인가?

### Sol

#### 1. 연구배경 (문제 상황 및 분석 목적)

본 연구는 흑색종(Melanoma) 환자의 조직생검에서 얻은 두 세포 집단인 **림프구 세포(Lymphocyte, $n_1=40$)** 와 **종양 세포(Tumor, $n_2=40$)** 의 직경(단위: $\mu\text{m}$) 데이터를 비교·분석한다. 주요 목적은 림프구 세포의 크기가 종양 세포의 크기보다 통계적으로 유의하게 작다고 볼 수 있는지 $5\%$ 유의수준에서 가설검정을 수행하는 것이다.

#### 2. 가설설정 (귀무가설/대립가설)

`1` 통계적 모형

 $$y_{ij} = \mu + \tau_i + \varepsilon_{ij},\quad (i = 1: \text{Lymphocyte},\,\,2: \text{Tumor}; \,\,j = 1, \dots, 40)$$

`2` 모평균 정의

- $\mu_1$: 림프구 세포 크기(직경)의 모평균
- $\mu_2$: 종양 세포 크기(직경)의 모평균

`3` 검정가설 (단측검정, One-sided Test)

- 귀무가설 ($H_0$): $\mu_1 \ge \mu_2$ (림프구 세포의 평균 크기가 종양 세포의 평균 크기보다 크거나 같다)
- 대립가설 ($H_1$): $\mu_1 < \mu_2$ (림프구 세포의 평균 크기가 종양 세포의 평균 크기보다 작다)

#### 3. EDA (자료특성 파악)

두 집단의 기술통계량 및 분포 특성은 다음과 같다.

| 세포 종류 | 표본수 (n) | 평균 (μm) | 표준편차 (s) | 분산 (s²) | 중앙값 (μm) | 사분위수범위 (IQR) |
|:---|---:|---:|---:|---:|---:|---:|
| 림프구 (Lymphocyte) | 40 | 6.950 | 1.596 | 2.547 | 7.100 | 2.625 |
| 종양 (Tumor) | 40 | 17.908 | 2.873 | 8.253 | 17.900 | 3.475 |


<br>

`1` 정규성 검정 (Shapiro-Wilk Test)

- 림프구 세포: $W = 0.9777, \; p = 0.6046$ ($p > 0.05$로 정규성 만족)
- 종양 세포: $W = 0.9805, \; p = 0.7092$ ($p > 0.05$로 정규성 만족)

`2` 등분산성 검정 (Folded F Test)

- $F = \frac{s_{\text{Tumor}}^2}{s_{\text{Lymphocyte}}^2} = \frac{8.2525}{2.5467} = 3.2405 \; (df_1 = 39, df_2 = 39)$

- $p\text{-value} = 0.0004$

- 유의수준 $0.05$에서 $p < 0.05$이므로 **"두 집단의 모분산이 같다"는 등분산 가정이 기각(이분산성 확인)된다.**

#### 4. 분석/모형적합

`1` 두 독립집단 데이터가 정규성 가정은 만족하지만 등분산 가정이 기각되었으므로,

`2` **Satterthwaite 근사법을 적용한 이분산 독립 2표본 t-검정(Satterthwaite/Welch Two-sample t-test)을 선택** 

`3` 보완적으로 비모수 검정인 Wilcoxon 순위합 검정도 병행 확인

#### 5. 결과해석 - 적합결과 및 검정통계량 해석

SAS 프로그램에서 `PROC TTEST` 산출 결과는 다음과 같다.

`1` **등분산 F-검정 (Equality of Variances)**
- $F = 3.24, \; df=(39, 39), \; p = 0.0004$ (이분산성)

`2` Satterthwaite t-검정 (SIDES=L, 이분산 적용)

- 검정통계량 $t = -21.09$
- 근사 자유도 $df = 60.98$
- 단측 $p\text{-value} < 0.0001$ ($5.28 \times 10^{-30}$)

`3` 비모수 Wilcoxon 순위합 검정 결과
- 정규근사 $Z = -7.66, \; p < 0.0001$

#### 6. 결론

::: {.panel-tabset}

##### 2-(1)

`1` 독립 2-표본 t-검정을 이용하여 림프구 세포와 종양 세포의 크기(직경) 평균을 비교한 결과, 

`2` 림프구 세포($n_1=40$)의 평균은 $6.95\,\mu\text{m}\,(SD=1.60)$, 종양 세포($n_2=40$)의 평균은 $17.91\,\mu\text{m}\,(SD=2.87)$이었다.

`3` 등분산성 검정 결과 $F = 3.24, p = 0.0004$로 등분산 가정이 기각되어 Satterthwaite 근사법을 적용.

`4`  검정 결과 $t(60.98) = -21.09$, 단측 $p < 0.0001$로 통계적으로 매우 유의한 차이가 있었음.

`5` 따라서 유의수준 5%에서 림프구 세포의 **평균** 크기가 종양 세포의 **평균** 크기보다 **통계적으로 유의하게 작다고 결론지을 수 있다.**

##### 2-(2)

독립 2표본 t-검정(Independent two-sample t-test)을 수행하기 위해 필요한 기본 가정은 다음과 같다

`1` **독립성 가정 (Independence)**: 두 집단(림프구 세포, 종양 세포) 내의 관측치들이 서로 독립적이어야 하며, 두 집단 간에도 서로 독립적으로 추출되어야 한다.

`2` **정규성 가정 (Normality)**: 각 집단에서 측정된 반응변수(세포 크기)의 모집단 분포가 정규분포를 따라야 한다
- (본 데이터는 $n_1=40, n_2=40$으로 충분히 크고 Shapiro-Wilk 검정에서 $p > 0.05$로 정규성을 만족함)

`3` **등분산성 가정 (Homogeneity of Variance)**: 두 모집단의 분산이 동일해야 한다는 가정이다[4].
* *참고*: 등분산성 가정이 성립하면 **합동분산(Pooled) t-검정**을 사용하지만,
* 본 데이터처럼 등분산 가정이 기각되는 경우에는 자유도를 보정한 **Satterthwaite(Welch) t-검정**을 적용하여 분석해야 한다
