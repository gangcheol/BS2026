## (1)

림프구세포의 크기가 조양세포의 크기보다 작다고 볼 수 있는가? 5% 유의수준으로 SAS를 이용하여 가설 검정하라. 

## (2)

위의 검정을 하기 위해 필요한 가정은 무엇인가?

### Sol

#### 1. 연구배경 (문제 상황 및 분석 목적)

`1` 본 연구는 흑색종(Melanoma) 환자의 조직생검에서 얻은 두 세포 집단인 **림프구 세포(Lymphocyte, $n_1=40$)** 와 **종양 세포(Tumor, $n_2=40$)** 의 직경(단위: $\mu\text{m}$) 데이터를 비교·분석한다.

`2` 주요 목적은 림프구 세포의 크기가 종양 세포의 크기보다 통계적으로 유의하게 작다고 볼 수 있는지 $5\%$ 유의수준에서 가설검정을 수행하는 것이다.

#### 2. 가설설정 (귀무가설/대립가설)

`1` 통계적 모형

 $$y_{ij} = \mu + \tau_i + \varepsilon_{ij},\quad (i = 1: \text{Lymphocyte},\,\,2: \text{Tumor}; \,\,j = 1, \dots, 40)$$

<details>
    <summary>기호 설명</summary>

- $y_{ij}$ (반응변수 / 관측값)
    - $i$번째 세포 집단(1: 림프구, 2: 종양)에 속한 $j$번째 개별 세포의 크기(직경, $\mu\text{m}$)
 
- $\mu$ (전체평균, Overall Mean)
    - 두 세포 집단 전체를 아우르는 기준 모평균값

- $\tau_i$ (집단효과, Group Effect): $i$번째 세포 종류가 세포 크기에 미치는 고유한 효과(처리 효과)
    - $\tau_1$: 림프구 세포(Lymphocyte) 집단의 효과 ($\mu_1 = \mu + \tau_1$)
    - $\tau_2$: 종양 세포(Tumor) 집단의 효과 ($\mu_2 = \mu + \tau_2$)

- $\varepsilon_{ij}$ (오차항, Random Error): 모형으로 설명되지 않는 $i$번째 집단 내 $j$번째 관측치의 확률적 오차
    - *(가정: 오차항은 서로 독립이고 평균이 0인 정규분포 $\varepsilon_{ij} \sim N(0, \sigma_i^2)$를 따름)*

- $i$ (집단 인덱스): 세포의 종류를 나타내는 첨자 ($i = 1$: 림프구 세포, $i = 2$: 종양 세포)

- $j$ (관측치 인덱스): 각 세포 집단 내 개별 세포 관측치의 번호를 나타내는 첨자 ($j = 1, 2, \dots, 40$)

</details>

`2` 모평균 정의

- $\mu_1$: 림프구 세포 크기(직경)의 모평균
- $\mu_2$: 종양 세포 크기(직경)의 모평균

`3` 검정가설 (단측검정, One-sided Test)

- 귀무가설 ($H_0$): $\mu_1 \ge \mu_2$ (림프구 세포의 평균 크기가 종양 세포의 평균 크기보다 크거나 같다)
- 대립가설 ($H_1$): $\mu_1 < \mu_2$ (림프구 세포의 평균 크기가 종양 세포의 평균 크기보다 작다)

#### 3. EDA (자료특성 파악)

두 집단의 기술통계량 및 분포 특성은 다음과 같다.

| 세포 종류 | 표본수 (n) | 평균 (μm) | 표준편차 (s) | 분산 (s²) | 중앙값 (μm) | 사분위수범위 (IQR) |
|:---|---:|---:|---:|---:|---:|---:|
| 림프구 (Lymphocyte) | 40 | 6.950 | 1.596 | 2.547 | 7.100 | 2.625 |
| 종양 (Tumor) | 40 | 17.908 | 2.873 | 8.253 | 17.900 | 3.475 |


<br>

`1` 정규성 검정 (Shapiro-Wilk Test)

<details>
    <summary>정규성 검정 코드</summary>

```python
/* 1. EDA 및 정규성 검정 (Shapiro-Wilk) */
PROC SORT DATA=ex3_2;
    BY group;
RUN;

PROC UNIVARIATE DATA=ex3_2 NORMAL PLOT;
    BY group;
    VAR x;
RUN;

```

</details>

- 림프구 세포: $W = 0.9777, \; p = 0.6046$ ($p > 0.05$로 정규성 만족)
- 종양 세포: $W = 0.9805, \; p = 0.7092$ ($p > 0.05$로 정규성 만족)

`2` 등분산성 검정 (Folded F Test)

<details>
    <summary> 등분산성 및 평균비교 검정 코드</summary>

```python
PROC TTEST DATA=ex3_2 SIDES=L ALPHA=0.05;
    CLASS group;
    VAR x;
RUN;
```

</details>

- $F = \frac{s_{\text{Tumor}}^2}{s_{\text{Lymphocyte}}^2} = \frac{8.2525}{2.5467} = 3.2405 \; (df_1 = 39, df_2 = 39)$

- $p\text{-value} = 0.0004$

- 유의수준 $0.05$에서 $p < 0.05$이므로 **"두 집단의 모분산이 같다"는 등분산 가정이 기각(이분산성 확인)된다.**

#### 4. 분석/모형적합

`1` 두 독립집단 데이터가 정규성 가정은 만족하지만 등분산 가정이 기각되었으므로,

`2` **Satterthwaite 근사법을 적용한 이분산 독립 2표본 t-검정(Satterthwaite/Welch Two-sample t-test)을 선택** 

`3` 보완적으로 비모수 검정인 Wilcoxon 순위합 검정도 병행 확인 $\to$ (소표본이므로 `Robust함을 확보하기 위함.`)

#### 5. 결과해석 - 적합결과 및 검정통계량 해석

SAS 프로그램에서 `PROC TTEST` 산출 결과는 다음과 같다.

`1` **등분산 F-검정 (Equality of Variances)**
- $F = 3.24, \; df=(39, 39), \; p = 0.0004$ (이분산성)

`2` Satterthwaite t-검정 (SIDES=L, 이분산 적용)

$$t = \frac{(\bar{X} - \bar{Y}) - \delta}{\sqrt{s_1^2/n_1 + s_2^2/n_2}}, \qquad \nu = \frac{(w_1 + w_2)^2}{w_1^2/(n_1 - 1) + w_2^2/(n_2 - 1)}, \qquad w_i = \frac{s_i^2}{n_i}$$

<details>
    <summary>수식 대입 결과</summary>

$$w_1 = \frac{1.596^2}{40} = 0.0637, \quad w_2 = \frac{2.873^2}{40} = 0.2063 \quad \Rightarrow \quad \nu = \frac{(0.0637 + 0.2063)^2}{0.0637^2/39 + 0.2063^2/39} \approx 60.98$$ $$t = \frac{6.950 - 17.908}{\sqrt{0.0637 + 0.2063}} \approx -21.09$$ **결론**: 귀무가설 기각 ($p < 0.0001$) — 림프구 세포의 크기가 종양 세포보다 통계적으로 유의하게 작음. *(참고: 자유도가 60.98로 정수가 아닌 이유는 두 표본의 분산 차이와 정보량을 비대칭적으로 반영하기 때문)*

</details>

`3` 비모수 Wilcoxon 순위합 검정 결과

<details>
    <summary>Wilcoxon 순위합 검정 코드</summary>

```python
/* 3. 비모수 검정 (Wilcoxon 순위합 검정 / Mann-Whitney U test) */
PROC NPAR1WAY DATA=ex3_2 WILCOXON;
    CLASS group;
    VAR x;
RUN;
```

</details>

`1` 본 자료는 정규성 가정을 만족하나 표본의 수가 적어  `Robust`함을 검증하기 위해 정규성 가정이 깨지고, 이분산일 경우 적용하는 `Wilcoxon 순위합 검정`을 함께 살펴봄.

`2` 검정결과 정규근사 $Z = -7.66, \; p < 0.0001$으로 단측 $p$-값이 유의수준 $\alpha = 0.05$보다 현저히 작으므로($p < 0.0001$), 귀무가설 $H_0$를 강력히 기각.

`3` **즉, 림프구 세포의 평균 크기가 종양 세포의 평균 크기보다 전반적으로 더 낮은 경향이 있었다.**

#### 6. 결론

::: {.panel-tabset}

##### 2-(1)

`1` 독립 2-표본 t-검정을 이용하여 림프구 세포와 종양 세포의 크기(직경) 평균을 비교한 결과, 

`2` 림프구 세포($n_1=40$)의 평균은 $6.95\,\mu\text{m}\,(SD=1.60)$, 종양 세포($n_2=40$)의 평균은 $17.91\,\mu\text{m}\,(SD=2.87)$이었다.

`3` 등분산성 검정 결과 $F = 3.24, p = 0.0004$로 등분산 가정이 기각되어 Satterthwaite 근사법을 적용.

`4`  검정 결과 $t(60.98) = -21.09$, 단측 $p < 0.0001$로 통계적으로 매우 유의한 차이가 있었음.

`5` 따라서 유의수준 5%에서 림프구 세포의 **평균** 크기가 종양 세포의 **평균** 크기보다 **통계적으로 유의하게 작다고 결론지을 수 있다.**

##### 2-(2)

독립 2표본 t-검정(Independent two-sample t-test)을 수행하기 위해 필요한 기본 가정은 다음과 같다

`1` **독립성 가정 (Independence)**: 두 집단(림프구 세포, 종양 세포) 내의 관측치들이 서로 독립적이어야 하며, 두 집단 간에도 서로 독립적으로 추출되어야 한다.

`2` **정규성 가정 (Normality)**: 각 집단에서 측정된 반응변수(세포 크기)의 모집단 분포가 정규분포를 따라야 한다
- (본 데이터는 $n_1=40, n_2=40$으로 충분히 크고 Shapiro-Wilk 검정에서 $p > 0.05$로 정규성을 만족함)

`3` **등분산성 가정 (Homogeneity of Variance)**: 두 모집단의 분산이 동일해야 한다는 가정이다[4].
* *참고*: 등분산성 가정이 성립하면 **합동분산(Pooled) t-검정**을 사용하지만,
* 본 데이터처럼 등분산 가정이 기각되는 경우에는 자유도를 보정한 **Satterthwaite(Welch) t-검정**을 적용하여 분석해야 한다

:::

## (1)

교육을 받기 전과 받은 후의 차이가 있는지를 유의수준 5%하에서 검정하여라.

## (2)

점수 차이의 모평균에 대한 95% 신뢰구간을 구하여라.

## (3)

(1)과 (2)사이의 연관성에 대하여 설명하여라.

### sol

#### 1. 연구배경

`1` 본 연구는 정신지체아를 둔 11명의 부모($n=11$)를 대상으로 자립심 향상 교육 실시 전(사전 점수, Before)과 실시 후(사후 점수, After)의 지식 습득 점수를 비교·분석한다.

`2` 동일한 연구 대상으로부터 처방(교육) 전후로 연속형 반응변수를 반복 측정한 짝지어진 자료(Paired data)이므로,

`3` 교육을 받기 전과 받은 후 지식 습득 점수에 통계적으로 유의한 차이가 있는지를 $5\%$ 유의수준에서 검정하고,

`4` 점수 차이 모평균의 $95\%$ 신뢰구간을 산출하는 것을 목적으로 한다.

#### 2. 가설설정

`1` 통계적 모형

$$d_i = y_{\text{after}, i} - y_{\text{before}, i} \quad (i = 1, 2, \dots, 11), \quad d_1, d_2, \dots, d_{11} \overset{iid}{\sim} N(\mu_d, \sigma_d^2)$$

<details>
    <summary>기호 설명</summary>

- $d_i$: $i$번째 부모의 교육 전후 지식 습득 점수 차이
- $y_{\text{after}, i}$: 교육 후 지식 습득 점수
- $y_{\text{before}, i}$: 교육 전 지식 습득 점수
- $\mu_d$: 교육 전후 점수 차이($\text{사후 점수} - \text{사전 점수}$)의 모평균
- $\sigma_d^{2}$: 교육 전후 점수 차이의 모분산

</details>

`2` 검정가설(양측검정, Two-sided Test)

- 귀무가설 ($H_0$): $\mu_d = 0$
    - 교육 전과 후의 지식 습득 점수 `평균에 차이가 없다`
- 대립가설 ($H_A$): $\mu_d \neq 0$
    - 교육 전과 후의 지식 습득 점수 `평균에 차이가 있다`

#### 3. EDA

11명 부모의 사전, 사후, 차이값에 대한 기술통계량은 다음과 같다.

| 변수 | 표본수($n$) | 평균 | 표준편차 ($s$) | 표준오차 ($SE$) | 중앙값 | 사분위수범위 (IQR) |
| :--- | :---: | :---: | :---: | :---: | :---: | :---: |
| **사전 점수 (Before)** | 11 | 81.909 | 17.067 | 5.146 | 84.000 | 23.000 |
| **사후 점수 (After)** | 11 | 84.000 | 16.273 | 4.906 | 83.000 | 22.000 |
| **점수 차이 ($d = \text{After} - \text{Before}$)** | 11 | 2.091 | 4.277 | 1.290 | 2.000 | 7.000 |

`1` **차이값(**$d$**)의 정규성 검정 (Shapiro-Wilk Test)**
- $W = 0.9170, \; p = 0.2941$
- $p > 0.05$이므로 차이값 분포가 정규성을 만족한다는 가정을 충족

::: {.panel-tabset}

##### before

##### after

#### 4. 분석/모형적합

`1` 차이값 $d$에 대한 정규성이 충족하므로 **모수적** 방법인 대응표본 `t-검정(Paired t-test)`을 적용

<details>
    <summary>SAS 실행 코드</summary>

```SAS
/* PROC TTEST의 PAIRED 문 사용*/
PROC TTEST DATA=ex3_4 ALPHA=0.05;
	PAIRED after*before; /* after - before 차이에 대한 t-검정 및 95% 신뢰구간 산출 */
RUN;
```

</details>

#### 5. 결과해석

`1` SAS `PROC TTEST` 산출 결과는 다음과 같다.

- 점수 차이 평균 ($\bar{d}$): $2.0909$점
- 표준오차 ($SE_{\bar{d}}$): $1.2895$점
- 자유도 ($df$): $n - 1 = 10$
- 검정통계량 ($t$): $t = \frac{\bar{d} - 0}{s_d / \sqrt{n}} = \frac{2.0909}{1.2895} = 1.6215 \approx 1.62$
- 양측 유의확률 ($p\text{-value}$): $p = 0.1360$
- 점수 차이 모평균($\mu_d$)에 대한 95% 신뢰구간:
    - $\bar{d} \pm t_{0.025, 10} \times SE_{\bar{d}} = 2.0909 \pm (2.2281 \times 1.2895) = [-0.7823, ; 4.9641]$

#### 6. 결론

::: {.panel-tabset}

##### 4-(1)

`1` 유의수준 $\alpha = 0.05$하에서 대응표본 t-검정을 실시한 결과, 검정통계량 $t(10) = 1.62$, 양측 $p\text{-value} = 0.1360$으로 나타남.

`2` 따라서 $p$-값이 유의수준 $0.05$보다 크므로($p > 0.05$) 귀무가설 $H_0$를 기각할 수 없다.

`3` **즉, 가 교육 전과 후 부모들의 지식 습득 점수에는 통계적으로 유의한 차이가 있다고 볼 수 없다.**

##### 4-(2)

`1` $d$(점수차이)의 모평균 $\mu_d$에 대한 95% 신뢰구간은  $[-0.7823, \; 4.9641]$ (약 $[-0.78, \; 4.96]$점)이다.

##### 4-(3)

`1` **(1)의 검정 결과**: 귀무가설 $H_0: \mu_d = 0$을 기각하지 못함 (차이가 유의하지 않음).

`2` **(2)의 신뢰구간**: 산출된 95% 신뢰구간 $[-0.7823, 4.9641]$ 내에 귀무가설의 기준값인 '0'이 포함되어 있다.

`3` **결론**:
    
-  귀무가설 $\mu_d = 0$을 유의수준 5%에서 기각할 수 없음을 의미하므로,
-  **(1)의 가설검정 결과와 (2)의 신뢰구간 결과는 완전하게 일치하고 연관**되어 있음을 의미한다!

### sol

#### 1. 연구배경

`1` 생활사건이 유발하는 심리적 스트레스가 면역기능에 영향을 미치는지 밝히는 것은 심신의학(Psychosomatic Medicine)의 핵심 연구 주제이다

`2` 본 연구는 전이성 폐암(Metastatic lung cancer) 환자 36명($N=36$)을 대상으로 최근 생활 스트레스 총량을 반영하는 SRRS(Social Readjustment Rating Scale) 점수를 측정하여

`3` 저($\le 54$), 중($55 \sim 99$), 고($\ge 100$) 세 집단(각 $n_i = 12$)으로 분류하고, 면역기능 지표인 NK 세포활동(Natural Killer cell activity) 수치를 측정하였다.

**연구질문**: **SRRS 스트레스 수준(저, 중, 고)에 따라 환자들의 NK 세포활동 평균에 차이가 있는가? 차이가 있다면 어느 집단 사이에 존재하는가?**

#### 2. 가설설정 (소문제(1) 답변)

`1` 통계적 모형

$$y_{ij} = \mu + \tau_i + \varepsilon_{ij} \quad (i = 1: \text{저}, \; 2: \text{중}, \; 3: \text{고}; \quad j = 1, \dots, 12), \quad \varepsilon_{ij} \overset{iid}{\sim} N(0, \sigma^2)$$

`2`  가설설정

$$H_0: \mu_1 = \mu_2 = \mu_3 \quad \text{vs} \quad H_1: \text{not }H_0$$

- $H_0$: 세 스트레스 집단의 `NK 세포활동 모평균이 모두 같다`.

- $H_1$: 적어도 한 집단의 `NK 세포활동 모평균은 다른 집단과 다르다`

#### 3. EDA

`1` **Shapiro-Wilk 정규성 검정**: 세 집단 모두 유의수준 5%에서 귀무가설(정규분포를 따른다)을 기각하지 못하여 **정규성 가정을 충족**

`2` 등분산성 검정 (Levene / Bartlett Test)
- 저 집단의 큰 분산($s^2 = 720.43$)으로 인해 등분산 가정이 기각되었다.
- 따라서 `표준 일원분산분석표`와 함께 등분산 위반에 강건한 **Welch 분산분석** 및 비모수 **Kruskal-Wallis 검정**을 교차 검증으로 제시한다.

##### EDA 분석 코드

```python
/******************************************************************
 * 1. 세 그룹 오버레이 히스토그램 (단일 그래프)
 ******************************************************************/
PROC SGPLOT DATA=ex3_5;
    HISTOGRAM activity / GROUP=group TRANSPARENCY=0.4; /* 그룹별 색상 구분 및 반투명 처리 */
    DENSITY activity / GROUP=group TYPE=NORMAL;        /* 그룹별 정규곡선 오버레이 */
    XAXIS LABEL="NK 세포활동 수치";
    YAXIS LABEL="비율 (%)";
    TITLE "SRRS 그룹별 NK 세포활동 오버레이 히스토그램";
RUN;

/******************************************************************
 * 2. 그룹별 색상 구분 상자그림 (단일 그래프)
 ******************************************************************/
/* CATEGORY=group GROUP=group 옵션을 함께 지정하면, x축에 세 그룹이 배치되면서 각각 고유한 색상(범례 포함)으로 채워집니다. */
PROC SGPLOT DATA=ex3_5;
    VBOX activity / CATEGORY=group GROUP=group; /* 그룹별 색상 구분 상자그림 */
    XAXIS LABEL="SRRS 스트레스 수준";
    YAXIS LABEL="NK 세포활동 수치" GRID;
    TITLE "SRRS 그룹별 NK 세포활동 상자그림";
RUN;

/******************************************************************
 * 3. 오버레이 Q-Q 플롯 (단일 그래프)
 ******************************************************************/
/* 1단계: PROC RANK를 통한 그룹별 이론적 정규 분위수(z_score) 산출 */
PROC SORT DATA=ex3_5;
    BY group;
RUN;

PROC RANK DATA=ex3_5 NORMAL=BLOM OUT=ex3_5_qq;
    BY group;
    VAR activity;
    RANKS z_score;
RUN;

/* 2단계: 단일 그래프에 세 그룹의 Q-Q 관측점 및 기준선 오버레이 */
PROC SGPLOT DATA=ex3_5_qq;
    SCATTER X=z_score Y=activity / GROUP=group;               /* 그룹별 색상 산점도 */
    REG X=z_score Y=activity / GROUP=group NOMARKERS;         /* 그룹별 기준 직선 */
    XAXIS LABEL="이론적 정규 분위수 (Theoretical Quantiles)";
    YAXIS LABEL="관측된 NK 세포활동 수치 (Sample Quantiles)" GRID;
    TITLE "SRRS 그룹별 NK 세포활동 오버레이 Q-Q 플롯";
RUN;

/******************************************************************
 * 참고: 1행 3열 가로 배치 패널 Q-Q 플롯
 ******************************************************************/
/* Q-Q 플롯을 한 그래프에 겹치는 것보다 가로 1줄로 나누어 보고 싶을 때 사용 */
PROC SGPANEL DATA=ex3_5_qq;
    PANELBY group / ROWS=1 COLUMNS=3;
    SCATTER X=z_score Y=activity / GROUP=group;
    REG X=z_score Y=activity / GROUP=group NOMARKERS;
    COLAXIS LABEL="이론적 분위수";
    ROWAXIS LABEL="관측 수치";
RUN;
```

:::

#### 4. 분석/모형적합 & 결과해석 (소문제(2), (3)번 답변)

::: {.panel-tabset}

##### **ANOVA Table**

| 변동요인 (Source) | 제곱합 (SS) | 자유도 (df) | 평균제곱 (MS) | F 통계량 | p-value |
|---|---|---|---|---|---|
| 집단 간 (Between Groups / 처리) | 4,506.11 | 2 | 2,253.06 | 7.85 | 0.0016 |
| 집단 내 (Within Groups / 오차) | 9,470.72 | 33 | 286.99 | - | - |
| 합계 (Total) | 13,976.83 | 35 | - | - | - |

$$F = \frac{MSTr}{MSE} = \frac{2,253.06}{286.99} = 7.85 \quad (df_1 = 2, \; df_2 = 33)$$

`1` F 통계량 및 p-value: $F(2, 33) = 7.85, \; p = 0.0016 < 0.05$

`2` $p$-값이 유의수준 $\alpha = 0.05$보다 현저히 작으므로 귀무가설 $H_0$를 기각한다.

- 즉, SRRS 스트레스 수준에 따라 환자의 NK 세포활동 평균에는 통계적으로 유의한 차이가 존재한다.

##### Tukey HSD 다중비교

| 비교 집단 | 평균차 | $p\text{-adj}$ | 판정 |
|---|---|---|---|
| 저 집단 vs 중 집단 | 24.87 | **0.0029** | 통계적으로 유의함 |
| 저 집단 vs 고 집단 | 22.41 | **0.0075** | 통계적으로 유의함 |
| 중 집단 vs 고 집단 | -2.46 | **0.9329** | 유의한 차이 없음 |

##### 보완 검정

* **Welch 분산분석**: $F(2, 20.3) = 6.22, \; p = 0.0078 < 0.05$
* **Kruskal-Wallis 비모수 검정**: $\chi^2(2) = 9.77, \; p = 0.0076 < 0.05$

> **해석**: 등분산 가정을 완화하거나 배제한 대안적 통계 검정에서도 동일하게 유의수준 5%에서 집단 간 유의한 차이가 입증되어 분석 결론이 매우 안정적이다.

##### SAS 코드

```python
/* (1) 일원분산분석 + 등분산 검정 + Tukey 다중비교 */
PROC GLM DATA=nk;
 CLASS group;
 MODEL activity = group;
 MEANS group / HOVTEST=LEVENE TUKEY;
RUN;
QUIT;
/* (2) 정규성 점검: 잔차 저장 후 Shapiro-Wilk */
PROC GLM DATA=nk;
 CLASS group; MODEL activity = group;
 OUTPUT OUT=out R=resid;
RUN;
QUIT;
PROC UNIVARIATE DATA=out NORMAL;
 VAR resid;
RUN;
/* (3) 비모수 대안 */
PROC NPAR1WAY DATA=nk WILCOXON;
 CLASS group; VAR activity;
RUN;
```

:::

#### 5. 결론

`1` 일원분산분석(모형 $y_{ij} = \mu + \tau_i + \varepsilon_{ij}$)을 이용하여 SRRS 저·중·고 세 집단 환자의 NK 세포활동을 비교하였다.

`2` 세 집단의 평균은 각각 저 집단 $40.47 \pm 26.84$, 중 집단 $15.60 \pm 6.42$, **고 집단 $18.06 \pm 9.97$**이었다.

`3` 분산분석 결과 $F(2, 33) = 7.85, \; p = 0.0016$으로 유의수준 5%하에서 통계적으로 유의한 차이가 있었고,

`4` Tukey 다중비교 결과, 스트레스가 가장 낮은 '저' 집단의 NK 세포활동 수치가 '중' 집단($p = 0.0029$) 및 '고' 집단($p = 0.0075$)보다 통계적으로 유의하게 높았으며, '중' 집단과 '고' 집단 사이에는 유의한 차이가 없었다($p = 0.9329$).

`5` 따라서 생활 스트레스 수치가 낮은 환자일수록 면역기능 지표인 NK 세포활동 능력이 평균적으로 높게 유지됨이 통계적으로 유의미하다는 것을 입증했다.

식이요법과 운동방법이 체중 감량에 미치는 영향분석

## 1. 연구배경

다이어트 연구자가 20명의 참가지(`N=20`)를 대상으로 \*\*'운동방법 5가지(A, B, C, D, E)'\*\*와 \*\*'식이요법 4가지(Ⅰ, Ⅱ, Ⅲ, Ⅳ)'\*\*의 조합(총 20개 셀)에 각 1명씩 임의 배정하고, 3주간 처치 후 체중 감량(kg)을 측정하였다.

### 연구질문

`1` 운동방법에 따라 체중 감량 효과에 통계적으로 유의한 차이가 있는가?

`2` 식이요법에 따라 체중 감량 효과에 통계적으로 유의한 차이가 있는가?

`3` 차이가 존재한다면, 어느 운동방법 및 식이요법 수준 사이에서 유의한 차이가 나타나는가?

::: {.callout-important}

1. 셀(각 조합별)당 관측치가 1개(`n=1`)이므로 오차항과 교호작용항을 분리할 수 없다. 따라서 두 요인이 서로 독립적으로 작용한다는 \*\*가법모형(Additive Model)\*\*을 적용하며,

2. 분석 전 **Tukey 가법성 검정(Tukey's Test of Additivity)** 을 통해 교호작용 부재 가정을 검증한다.

:::
